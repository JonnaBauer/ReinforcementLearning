d)  Give an intuitive interpretation of the plotted value function.

In the plotted value function you can see that the value peaks at the upright equilibrium (angle = 0, $\dot{\theta} = 0$), where the system recieves the maximal reward, whereas states further away from the upright positon sharply drops in value, so the function captures the underlying cost-to-go. States far from the goal require significant control effort and time to swing up, resulting in lower value due to discounting and accumulated penalties. 

(e) Compare the learning curve and the resulting value function when using
γ = 0.5. What happens?

When using γ = 0.5, the learning curve drops more quickly and it's much smoother. This is because future rewards are heavily discounted which makes the regression problem smotther and easier for the neural network to fit.

The resulting value function becomes significantly flatter and compressed in magnitude, which is because the value function fails to properly accumulate long-term costs over extended trajectories, ending up underestimating the true cost-to-go for states that require a long sequence of actions to reach the goal.
